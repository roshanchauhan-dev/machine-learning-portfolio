# Linear Regression using sklearn
This notebook demonstrates basic linear regression.

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import fetch_openml # It is Used to download datasets from the OpenMl(online Source)
from sklearn.model_selection import train_test_split #Import function to spilt data into training and testing parts
from sklearn.preprocessing import StandardScaler # Make all values same range 
from sklearn.linear_model import LinearRegression # Machine learning model that learns and predicts values
from sklearn.metrics import r2_score # Used to check model performance(accuracy)
```

```python
# Fetch the data from the web
boston=fetch_openml(name='boston',version=1,as_frame=True)
df = boston.frame
df.head() 
# House price predication datasets
```

```python
# Separate into X (Features) and y (Target)
X = df.drop('MEDV',axis=1) # X -> all features(remove MEDV column because it is the target)
y=df['MEDV'] # y -> target variable (house price we want to predict)
```

```python
# Split into training and testing sets
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
# Make all features similar range
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train) #learn+scale
X_test=scaler.transform(X_test) # only scale
```

```python
regression = LinearRegression()
regression.fit(X_train,y_train)
predictions=regression.predict(X_test)
```

```python
score=r2_score(y_test,predictions)
print("Accuracy Score:",score)
# if score 1.0 is perfect,if 0.7+ then good,if 0.5 average,if <0.5 weak 
```

```python
import matplotlib.pyplot as plt
plt.scatter(y_test,predictions)
plt.xlabel("Actual Price")
plt.ylabel("predicted Price")
plt.title("Linear Regression Result")
plt.show()
```
